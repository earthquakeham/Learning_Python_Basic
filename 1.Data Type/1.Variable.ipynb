```python
# Various Data Type
intger = 10
float = 10.5
string = 'Hello'
datetime = '2024-11-22'
list = [1,2,3]
tuple = (1,2,3)
dictionary = {'name': 'Earthquakeham', 'age': 48}
set = {1,2,3}
boolean = True

print(type(intger))
print(type(float))
print(type(string))
print(type(datetime))
print(type(list))
print(type(tuple))
print(type(dictionary))
print(type(set))
print(type(boolean))
```

```python
# 변수는 값이 아닌 주소를 저장한다.
a = [1,2,3]
b = a
a[1] = 4
print(a)      # a에 저장된 값 출력
print(b)      # b에 저장된 값 출력
print(id(a))  # a에 할당된 주소 출력
print(id(b))  # b에 할당된 주소 출력
```

```python
# 값을 복사 하는 방법
a = [1,2,3]
b = a[:]      # a를 list로 만들어 b에 반영
a[1] = 9      # a 값 변경
print(a)      # a에 저장된 값 출력
print(b)      # b에 저장된 값 출력
print(id(a))  # a에 할당된 주소 출력
print(id(b))  # b에 할당된 주소 출력
```
