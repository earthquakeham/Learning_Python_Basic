```python
# Define String
text1 = "You're my friends."
text2 = 'You\'re my friends.'   # 여기서 \는 '를 문자열로 인식하게 한다.
text3 = "You're my friends. But they are not my friends"
text4 = "You're my friends. \nBut they are not my friends"
text5 = "Jane\\Mike are my friends. \tBut Joe is not my friends"
text6 = '''I like Jane
but Jane like Mike... ^^
So sad... '''

print(text1)
print(text2)
print(text3)
print(text4)
print(text5)
print(text6)
```

```python
# Indexing/Slicing
name = 'Earthquakeham'
print(name[3])
print(name[1:8])
print(name[:8])
print(name[:8:2])
print(name[::2])
print(name[-1])
print(name[2:-1])
print(name[::-2])
```

```python
# String 연산 "+"
P1 = "L3V343E06011_L3V343E06011AA0_3CIL02_LOI_21230_BLUE_C_20240605_213354.jpg"
G1 = "L3V343E06011_21231_3CIL02_LOI_BLUE_C_20240605_213456.JPG"

date_panel = [P1[-19:-11] + P1[-10:-4]]
date_glass = [G1[-19:-11] + G1[-10:-4]]

print(date_panel)
print(date_glass)
```

```python
# String 연산 "*"
print("=" * 50)
print("Hello" * 3)
print("=" * 50)
```

```python
# formatting 기본 사용법
number = 10
days = 'ten'
text1 = 'I ate %d apple for %s days' %(5, 'five')
text2 = 'I ate %d apple for %s days' %(number, days)
print(text1)
print(text2)
```

```python
# %s와 %r 차이
x = "hello\nworld"
print("String with repr : %r" %x)  # %r(repr)은 문자열 그대로 출력
print("String with str : %s" %x)  # %s(str)은 /n 인식
```

```python
# %s와 %10s 차이.... 10칸 공백
text3 = 'Jane lives %smiles from here'%(3)     # %s 공백 없음
text4 = 'Jane lives %10smiles from here'%(3)  # %10s 공백 있음
print(text3)
print(text4)
```

```python
text7 = 'She is %10s ...' % 'Jane'
text8 = 'She is %-10s ...' % 'Jane'
print(text7)
print(text8)
```

```python
# %f와 %0.4f 차이.... 소수점 자리수
text5 = 'Jane lives %fmiles from here'%(3.1415926)  # %f 소수점 6자리
text6 = 'Jane lives %0.4fmiles from here'%(3.1415926)  # %0.4f 소수점 4자리
print(text5)
print(text6)
```

```python
# f-string

# f를 사용하여 f-string을 사용할 것을 선언 후 {}안에 변수 입력
old_full_path = "/path/to/old_file.txt"
new_full_path = "/path/to/new_file.txt"
message = f"Renamed: {old_full_path} -> {new_full_path}"
print(message)

# {}안에서 변수의 표현식도 가능함
x = 5
y = 10
print(f"The sum of {x} and {y} is {x + y}.")

# dictionary 응용
d = {'name':'Earthquakeham', 'age':48}
print(f'My name is {d["name"]}. I am {d["age"]} years old.')
```

```python
# function 응용1
name = 'Earthquakeham'
print(name.lower())  # 소문자로 변환
print(name.upper())  # 대문자로 변환
print(name.count('h'))  # 문자열에서 특정 문자의 개수 반환
print(name.find('r'))  # 문자열에서 특정 문자의 위치 반환
print(name.index('r'))  # 문자열에서 특정 문자의 위치 반환
print(name.lstrip())  # 왼쪽 공백 제거
print(name.rstrip())  # 오른쪽 공백 제거
print(name.strip())  # 양쪽 공백 제거
print(name.replace('E', 'F'))  # 문자열에서 특정 문자를 다른 문자로 변환
print(name.swapcase())  # 대소문자 변환

name2 = 'Hello   World  Python   Programming'
print(name2.split())  # 모든 종류의 공백(스페이스, 탭, 줄바꿈)을 기준으로 분리
print(name2.split(' '))  # 정확히 스페이스 한 칸만을 기준으로 분리
print(name2.split(' ', 2))  # 최대 2번만 분할하여 3개의 요소를 가진 리스트 반환
print(name2.splitlines())  # 줄바꿈을 기준으로 문자열을 나누어 리스트로 반환
```

```python
# function 응용2

# 1. 기본적인 join 사용
words = ['Python', 'is', 'awesome']
print(''.join(words))  # words 리스트의 각 요소를 빈 문자열로 연결
print(' '.join(words))  # words 리스트의 각 요소를 공백으로 연결
print('-'.join(words))  # words 리스트의 각 요소를 '-'로 연결

# 2. 줄바꿈으로 join
lines = ['첫째줄', '둘째줄', '셋째줄']
print('\n'.join(lines))  # lines 리스트의 각 요소를 줄바꿈으로 연결

# 3. 문자열에 대한 join
print(','.join('Python'))  # 'Python' 문자열의 각 문자를 ','로 연결
```
