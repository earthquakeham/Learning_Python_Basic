```python
# Indexing, Slicing
names = ['Mason','Harris','Jane','Korey','Bob','Kelly']

print("Example of slicing[0:5] :", names[0:5])
print("Example of slicing[0:5:1] :", names[0:5:1])
print("Example of slicing[0:5:2] :", names[0:5:2])
```

```python
# Operation
a = [1, 2, 3]
b = [4, 5, 6]
print("Example of a + b :", a + b)
print("Example of a * 3 :", a * 3)
print("Example of b * 5 :", b * 5)
```

```python
# "del"

names = ['Mason','Harris','Jane','Korey','Bob','Kelly']

del names[4:6] # names 변수에 있는 list에서 [4:6]으로 slicing 하여 제거
print("Example of del :", names)
length = len(names) #
print("Example of len :", length)
```

```python
# function
names = ['Mason','Harris','Jane','Korey','Kelly']

names.append('Bob') # 기존 list에 새로운 항목을 추가하는 함수
print("Example of append : ", names)
names.insert(2, 'Tom') # insert 함수는 기존 list에 특정 위치에 새로운 항목을 추가하는 함수
print("Example of insert : ", names)
names.extend(['LK', 'JF']) # extend 함수는 기존 list 끝에 새로운 항목을 추가하는 함수
print("Example of extend : ", names)
names.sort() # sort 함수는 기존 list를 정렬하는 함수
print("Example of sort : ", names)
names.reverse() # reverse 함수는 기존 list를 역순으로 정렬하는 함수
print("Example of reverse : ", names)
exercise1 = names.index('Harris') # index 함수는 기존 list에서 특정 항목의 위치를 반환하는 함수
print("Example of index : ", exercise1)
names.remove('Mason') # remove 함수는 기존 list에서 특정 항목을 제거하는 함수
print("Example of remove : ", names)
names.pop() # pop 함수는 기존 list에서 마지막 항목을 제거하는 함수
print("Example of pop : ", names)
exercise2 = names.count('LK') # count 함수는 기존 list에서 특정 항목의 개수를 반환하는 함수
print("Example of count : ", exercise2)
```
