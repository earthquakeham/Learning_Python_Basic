```python
# list vs tuple

list_sample = [1,2,3,4]
print("Example of list_sample :", list_sample)
list_sample[0] = 5              # list는 값 변경 가능
print("Example of list_sample :", list_sample)

# tuple
tuple_sample = (1,2,3,4)
print("Example of tuple_sample :", tuple_sample)
tuple_sample[0] = 5             # tuple은 값 변경 불가. Error 발생
print("Example of tuple_sample :", tuple_sample)
```

```python
# various tuple. 2가지 특이점이 있다.

a = ()
b = (1,)                         # 단지 1개의 요소만을 가질 때는 요소 뒤에 쉼표(,)를 반드시 붙여야 한다
c = (1, 2, 3)
d = 1, 2, 3                      # 괄호를 생략해도 무방하다
e = ('Life', 'is', 'too', 'short')
f = (1, 2, 'Life', 'is')
g = (1, 2, ['Life', 'is'])
h = (1, 2, ('Life', 'is'))
```

```python
# Operation

tuple1 = (1,2,3)
tuple2 = (4,5,6)
sum = tuple1 + tuple2
multiplication = tuple1 * 3
print("Example of sum :", sum)
print("Example of multiplication :", multiplication)
```
