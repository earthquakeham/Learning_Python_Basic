```python
# 변수 방식, list 방식, dictionary 방식 비교

# 변수 방식 Data
name1 = 'Mason'
Age1 = 26
Sex1 = 'male'
Hair1 = 'Black'
name2 = 'Jane'
Age2 = 21
Sex2 = "female"
Hair2 = 'Red'

# list 방식 Data
person1 = ['Mason', 26, 'male', 'Black'] 
person2 = ['Jane', 21, 'female', 'Red'] 

# dictionary 방식 Data
person1 = {'name':'Mason', 'Age':26, 'Sex':'male', 'Hair':'Black'}
person2 = {'name':'Harris', 'Age':21, 'Sex':'female', 'Hair':'Red'}
```

```python
# keys, values, items
person1 = {'name':'Mason', 'Age':26, 'Sex':'male', 'Hair':'Black'}

for key in person1.keys():
    print("Example of keys :", key)

for value in person1.values():
    print("Example of values :", value)

for key, value in person1.items():
    print("Example of items :", key, value)
```

```python
# add, delete, return
person1 = {'name':'Mason', 'Age':26, 'Hair':'Black'}
person2 = {'name':'Harris', 'Age':21, 'Hair':'Red'}

# add
person1['food'] = 'Pizza'
print("Example of add :", person1)
# partially delete key
del person1['food']
print("Example of partially delete key :", person1)
# fully delete key
person1.clear()
print("Example of fully delete key :", person1)
# value 반환
print(person2.get('name'))    # dic이름.get(key) 형식
print(person2['name'])        # dic이름[key] 형식
print(person2.get('name2'))   # 없는 key 호출할 경우 > 'none' 반환
print(person2['name2'])       # 없는 key 호출할 경우 > error
```

```python
# Indexing, Slicing
# Dictionary는 순서가 없기 때문에 인덱싱, 슬라이싱 불가능
# 하지만 key를 통해 value를 반환할 수 있다

person1 = {'name':'Mason', 'Age':26, 'Hair':'Black'}
print("Example of person1['name'] :", person1['name'])
print("Example of person1['Age'] :", person1['Age'])
print("Example of person1['Hair'] :", person1['Hair'])
```

```python
# function
person1 = {'name':'Mason', 'Age':26, 'Hair':'Black'}
print("Example of keys :", person1.keys()) # key를 list 형태로 반환
print("Example of values :", person1.values()) # value를 list 형태로 반환
print("Example of items :", person1.items()) # (key, value)를 list 형태로 반환
print("Example of get :", person1.get('name')) # key를 통해 value 반환
print("Example of clear :", person1.clear()) # 모든 key 삭제
```
