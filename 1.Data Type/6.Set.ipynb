```python
# Varioous set
set1 = set([1,2,3,4])       # set 형식1
set2 = {5,6,7,8}            # set 형식2
set3 = set("Hello")          # set 형식3
list1 = [1,2,2,3,4,3,5,2,5] # 중복된 값을 가지는 list

print("Example of 형식1 :", set1) 
print("Example of 형식2 :", set2)
print("Example of 형식3 :", set3)
print("Example of list :", list1)
print("Example of list to set :", set(list1))
```

```python
# Operation
set1 = set([1,2,3,4,5,6])
set2 = {4,5,6,7,8,9,10}

print("Example of union :", set1 | set2) # union
print("Example of intersection :", set1 & set2) # intersection
print("Example of difference :", set1 - set2) # difference
print("Example of difference :", set2 - set1) # difference 
```

```python
# add, update, remove
set1 = set([1,2,3,4,5,6])
set2 = {4,5,6,7,8,9,10}

set1.add(7)                # 한개 추가
print("Example of add :", set1)
set2.update([11,12,13,14]) # list 형태 추가
print("Example of update :", set2)
set2.remove(4)            # 특정 값 제거
print("Example of remove :", set2)
```
