```python
# 문자 인식을 통한 구분
boolsample = "boolean"

if boolsample:
    print('It is true')
else : 
    print('It is false')
```

```python
print('Example of "python" :', bool("python"))
print('Example of "" :', bool(""))
print('Example of "[1,2,3]" :', bool([1,2,3]))
print('Example of "[]" :', bool([]))
print('Example of "{}" :', bool({}))
print('Example of "()" :', bool(()))
print('Example of "1" :', bool(1))
print('Example of "0" :', bool(0))
print('Example of "None" :', bool(None))
```

```python
x = 0    # flase를 의미함
y = 1    # true를 의미함

print(bool(x or y))
print(bool(x and y))
print(bool(not x))
print(bool(x ^ y))    # ^ = nor
```
