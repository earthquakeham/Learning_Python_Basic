```python
# 주민등록번호 뒤자리 별표로 만드는 방법1(정규 표현식 미사용)

data = """
Choi 770120-2056811
Ham  770322-1168115
"""

result = []
for line in data.split("\n"):
    word_result = []
    for word in line.split(" "):
        if len(word) == 14 and word[:6].isdigit() and word[7:].isdigit():
            word = word[:6] + "-" + "*******"
        word_result.append(word)
    result.append(" ".join(word_result))
print("\n".join(result))
```

```python
# 주민등록번호 뒤자리 별표로 만드는 방법(정규 표현식 사용)
import re 

data = """
Choi 770120-2056811
Ham  770322-1168115
"""

pat = re.compile("(\d{6})[-]\d{7}")
print(pat.sub("\g<1>-*******", data))
```
