```python
# 산술 연산
x = 7 
y = 8

print(x+y)
print(x-y)
print(x*y)
print(x/y)
print(x//y)
print(x%y)
print(x**y)
```

```python
# 문자열 연산
text1 = 'I like'
text2 = ' python'
text3 = ' very much'

total_text1 = text1 + text2 + text3
total_text2 = text1 + "\n" + text2 + "\n" + text3*3
print(total_text1)
print(total_text2)
```

```python
# 논리 연산
x = 0    # flase를 의미함
y = 1    # true를 의미함

print(bool(x or y))
print(bool(x and y))
print(bool(not x))
print(bool(x ^ y))    # ^ = nor
```
