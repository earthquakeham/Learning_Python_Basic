```python
# 조건문 if sample1

distance = int(input('Input enemy distance from you>>>'))
if distance >= 200:
    print('Use gun!!!')
elif distance >= 50:
    print('Use arrow')
else:
    print('Use knife!!!')
```

```python
# 조건문 if sample2

price = int(input('Input price of TESLA($) : '))
if price >= 300:
    print('sell')
elif price >= 200:
    print('wait')
else:
    print('buy')
```

```python
# 조건문 if sample3

price_car1 = int(input('Input 1st car price :'))
price_car2 = int(input('Input 2nd car price :'))
price_total = price_car1 + price_car2
price_real = price_total
print('Total price :',price_total)
if price_total >= 200000:
    price_real = price_real * 0.5
elif price_total >= 150000:
    price_real = price_real * 0.75
else:
    price_real = price_real * 1

print('Real price : ',price_real)
```

```python
# elif concept
# if, else only

status = ["Hungry", "Not Hungry"]
money = 1

if "Not Hungry" in status:
    print("Do Exercise")
else:
    if money:
        print("Buy and eat something")
    else:
        print("Go back to home!")
```

```python
# elif concept
# if, elif, else

status = ["Hungry", "Not Hungry"]
money = 1

if "Not Hungry" in status:
    print("Do Exercise")
elif money:
    print("Buy and eat something")
else:
    print("Go back to home!")
```
