```python
# for

# 반복 횟수가 정해진 경우
for i in range(5):    # 5번 반복
    print(i)

# 시퀀스(리스트, 튜플 등)를 순회할 때
fruits = ['apple', 'banana', 'orange']
for fruit in fruits:    # 리스트의 요소 수만큼 반복
    print(fruit)
```

```python
# while

# 반복 횟수가 정해지지 않은 경우, 조건이 만족하는 동안 계속 반복
count = 0
while count < 5:    # 조건이 False가 될 때까지 반복
    print(count)
    count += 11
    

# 사용자 입력을 받을 때
user_input = ''
while user_input != 'quit':    # 'quit' 입력할 때까지 반복
    user_input = input('명령어를 입력하세요: ')
```
