```python
# list 이용
names = ['Mason','Harris','Kelly','Bob']
for name in names:
    if name == 'Mason':
        print(name + ' is captain')
    elif name == 'Harris':
        print(name + ' is member')
    else :
        print(name + ' is enemy')
```

```python
# range() 이용
for count in range(1,10,2):
    print(count, ' enemy detected')
```
