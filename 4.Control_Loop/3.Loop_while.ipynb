```python
# While... True... 이용
while True:
    num = int(input('1.Start / 2. Record / 3.End : '))

    if num == 1:
        print('Start Game')
    elif num == 2:
        print('Game Record')
    elif num == 3:
        print('End Game')
        break
    else:
        print('Input again')
```

```python
# cotinue & break & pass

# continue 문은 반복문(for 또는 while) 안에서 사용되며, 현재 반복을 건너뛰고 다음 반복을 계속 수행하도록 만든다. 반복문 안에서 특정 조건을 만족할 때, 그 조건에 해당하는 작업을 건너뛰고 싶을 때 continue를 사용.

for i in range(1, 6):
    if i == 3:
        continue
    print("continue", i)

# break
# break 문은 반복문을 완전히 종료시킨다. break가 호출되면 현재 반복을 포함한 전체 반복문이 종료되고, 반복문 다음의 코드가 실행된다.

for j in range(1, 6):
    if j == 3:
        break
    print("break", j)

# pass
# pass는 아무 동작도 하지 않는다. 문법적으로 문장이 필요한 곳에서 자리를 채우는 용도로 사용된다. 반복문 안에서 특정 조건을 만족할 때, 그 조건에 해당하는 작업을 건너뛰고 싶을 때 pass를 사용.

for k in range(5):
    if k == 2:
        pass
    print("pass",k)
```
