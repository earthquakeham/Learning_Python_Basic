```python
# 구구단 출력
for i in range(2,10):
    for j in range(1,9):
        print(i*j)
```

```python
# 사용자에게 자연수를 입력받아서 0부터 합계를 구하여 출력
num = int(input('Input number : '))
sum = 0
for i in range(1,num+1):
    sum = sum + i

print(sum)
```

```python
# 사용자에게 자연수를 입력받다 4가 입력되면 프로그램 종료

# 방법1. 
num = int(input('Input number : '))
while num != 4:
    num = int(input('Input number : '))
print('End program')

# 방법2. while... True...
num2 = int(input('Input number2 : '))
while True:
    num2 = int(input('Input number2 : '))
    if num2 == 4:
        break
print('End program')
```
