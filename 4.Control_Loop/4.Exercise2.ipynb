```python
# *
# **
# ***
# ****
# *****
print('1st star')
for i in range(1,6):
    print('*' * i)
```

```python
# *****
# ****
# ***
# **
# *
print('2nd star')
for i in range(1,6):
    print('*' * (6-i))
```

```python
#     *
#    **
#   ***
#  ****
# *****
print('3rd star')
for i in range(1,6):
    print((' '* (5-i)),('*'*i))
```

```python
# *****
#  ****
#   ***
#    **
#     *
print('4th star')
for i in range(5):
    print((' '* i),('*'*(5-i)))
```
