```python
# Return이 있는 함수
def cal(a,b,c):                  # def 함수_이름(매개변수):
    calculation = (a + b) * c    # 수행할_문장
    return calculation           # return 반환값
x = cal(3,4,5)                   # 함수_이름(인자)
print(x)                         # 반환값 출력
```

```python
# Return이 없는 함수
def printtext(text):             # def 함수_이름(매개변수):
    print(text)                  # 수행할_문장
printtext("no return.")          # 함수_이름(인자)
```

```python
# 매개변수가 없는 함수
import random
def randnumber():                 # def 함수_이름(매개변수):
    number = random.randint(1,50) # 수행할_문장
    return number                 # return 반환값
print(randnumber())               # 함수_이름(인자)
```

```python
# return, 매개변수 둘 다 없는 함수
def sayhi():                     # def 함수_이름(매개변수):
    print('Hi')                  # 수행할_문장
sayhi()                          # 함수_이름(인자)
```

```python
# name mangling
"""
더블 언더스코어로 시작하고 끝남 (__name__ 형식)
이는 일반적으로 클래스 내부에서 private 메서드를 나타낼 때 사용
외부에서의 직접 접근을 제한하는 용도
"""
def __randnumber__():             # def 함수_이름(매개변수):
    number = random.randint(1,50) # 수행할_문장
    return number                 # return 반환값
print(randnumber())               # 함수_이름(인자)
```
