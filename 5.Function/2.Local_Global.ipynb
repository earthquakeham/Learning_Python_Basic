```python
# Difference between local and global variable
a = 1
def test1(a):      # local a = 1, global a = 1
    a = a + 1      # local a = 2, global a = 1

test1(1)           # local a = 2, global a = 1
print(a)           # global a = 1
```

```python
# Local variable to Global variable1
b = 1
def test2(b):
    b = b + 10     # local b = 11, global b = 1
    return b       # local b = 11, global b = 1
b = test2(1)       # local b = 11, global b = 11
print(b)           # global b = 11

# Local variable to Global variable2
c = 1
def test3():
    global c        # global c = 1
    c = c + 1       # global c = 2
test3()             # global c = 2
print(c)            # global c = 2
```
