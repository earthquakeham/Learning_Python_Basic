```python
# 함수 사용하지 않고 파일 이동

import os
import shutil

source_dir = r"C:\Users\lilin\Desktop\Test_Before"
dest_dir = r"C:\Users\lilin\Desktop\Test_After"
min_size = 1000000

for file_name in os.listdir(source_dir):

    file_path = os.path.join(source_dir, file_name)
    file_size = os.path.getsize(file_path)

    if file_name.endswith('.exe'):
        if file_size > min_size:
            shutil.move(file_path, os.path.join(dest_dir, file_name))

print(f"디렉토리 {source_dir}가 {dest_dir}로 복사되었습니다.")
```

```python
# 함수 사용하여 파일 이동

import os
import shutil

def move_file(source_dir, dest_dir, extension, min_size):
    for file_name in os.listdir(source_dir):
        file_path = os.path.join(source_dir, file_name)
        file_size = os.path.getsize(file_path)

        if file_name.endswith(extension):
            if file_size > min_size:
                shutil.move(file_path, os.path.join(dest_dir, file_name))

source_dir = r"C:\Users\lilin\Desktop\Test_Before"
dest_dir = r"C:\Users\lilin\Desktop\Test_After"
min_size = 1000000

move_file(source_dir, dest_dir, ".exe", min_size)

print(f"디렉토리 {source_dir}가 {dest_dir}로 복사되었습니다.")
```
