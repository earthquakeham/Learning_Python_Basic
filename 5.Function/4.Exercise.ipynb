1. 1~50중 random하게 6개의 숫자를 생성
2. 6개의 숫자는 중복 될 수 있음

```python
# 1~50안에서 중복 가능하게 무작의 6개 숫자 뽑기 
import random

def randumnumber():
    number = random.randint(1,50)
    return number

accu_number = []

for i in range(6): # 0,1,2,3,4,5
    rand_number = randumnumber()
    accu_number.append(rand_number)
print(accu_number)
```

1. 1~50중 random하게 6개의 숫자를 생성
2. 6개의 숫자는 중복 되지 않음

```python
# 1~50안에서 중복 되지 않게 무작의 6개 숫자 뽑기 
import random

def randomnumber():
    number = random.randint(1,50)
    return number

accu_number = []
count = 0

while True:
    if count > 5:   #
        break
    if randomnumber() not in accu_number:
        accu_number.append(randomnumber())
        count = count + 1
print(accu_number)
```
