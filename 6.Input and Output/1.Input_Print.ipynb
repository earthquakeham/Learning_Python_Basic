```python
# print에서 간격
print('Life'+'is'+'to'+'short') # +는 문자열을 이어 붙임
print('Life' 'is' 'to' 'short') # 빈칸은 문자열을 이어 붙임
print('Life','is','to','short') # ,는 한칸 뛰어 쓰기
```

```python
# print 형식
for i in range(10):
    print(i, end=' ')      # 반복 실행 후 끝에 공백을 추가

for j in range(10):
    print(j, end='\n')     # 반복 실행 후 끝에 한줄 내려 쓰기 추가 > print(j) 와 동일한 결과
```
