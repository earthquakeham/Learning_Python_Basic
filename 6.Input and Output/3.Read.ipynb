```python
# readline... 한줄씩 불러오기
file_root = open('/Users/earthquakeham/Library/Mobile Documents/com~apple~CloudDocs/PYTHON/write_test2.txt', 'r')
while True:
    line = file_root.readline()  # 한줄씩 읽기
    if not line:                 # 더 이상 읽을 줄이 없으면 종료
        break
    print(line)
file_root.close()
```

```python
# readlines... 각 줄을 list 형태로 불러오기
file_root = open('/Users/earthquakeham/Library/Mobile Documents/com~apple~CloudDocs/PYTHON/write_test2.txt', 'r')
lines = file_root.readlines()      # 각 줄을 list 형태로 불러오기
for line in lines:                 # 각 줄을 출력
    print(line.replace('\n',' '))  # 줄 끝의 /n을 공란으로 변경
file_root.close()
```

```python
# read... 파일 통째로 불러오기
file_root = open('/Users/earthquakeham/Library/Mobile Documents/com~apple~CloudDocs/PYTHON/write_test2.txt', 'r')
data = file_root.read()              # 파일 통째로 불러오기
print(data)
file_root.close()
```
