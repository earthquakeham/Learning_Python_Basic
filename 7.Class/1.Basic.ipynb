```python
class hero_info:
    def __init__(self,name, age):
        self.name = name
        self.age = age
    def say_name(self):
        print(f'My name is {self.name}.')
    def say_age(self):
        print(f'I am {self.age} years old')

hero1 = hero_info('Batman', 26)
hero2 = hero_info('Spiderman', 28)

hero1.say_name()
hero1.say_age()
hero2.say_name()
hero2.say_age()
```
