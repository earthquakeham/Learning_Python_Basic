```python
# Class에 기능 추가. self 개념 적용
class jordan:                                # jordan 이라는 class 생성
    explain = 'WOW! It is NIKE jordan1!'     # jordan class의 "explain" 특성 정의
    def color(self,color):                   # jordan class의 "color" 특성 정의
        self.color = color    
        print(color,"color!")
    def price(self,price_origin,discount):   # jordan class의 "price" 특성 정의
        price_real = price_origin * (1-(discount/100))
        print('Real price($) :',price_real)

color_input = input("Input Color : ")
price_input = int(input("Input original price($) : "))
discount_input = int(input("Input discount(%) : "))

shose_type1 = jordan()                       # "shose_type1" object는 jordan class 사용함
print(shose_type1.explain)                   # "explain" 기능 사용
shose_type1.color(color_input)               # "color" 기능 사용
shose_type1.price(price_input,discount_input)# "price" 기능 사용
```

```python
# Class에 기능 추가. 'self 개념' 적용 + 'init 개념' 적용
class jordan:                                # jordan 이라는 class 생성
    explain = 'WOW! It is NIKE jordan1!'     # "explain" 특성 정의
    def __init__(self,color):                # __init__ 반영하여 "color" 특성 정의
        self.color = color    
        print(color,"color!")
    def price(self,price_origin,discount):   # jordan class의 "price" 특성 정의
        price_real = price_origin * (1-(discount/100))
        print('Real price($) :',price_real)

color_input = input("Input Color : ")    
price_input = int(input("Input original price($) : "))
discount_input = int(input("Input discount(%) : "))

shose_type1 = jordan(color_input)            # class에서 __init__이 있기 때문에 괄호 안에 초기값 입력
print(shose_type1.explain)                   # "explain" 기능 사용
shose_type1.price(price_input,discount_input)# "price" 기능 사용
```
