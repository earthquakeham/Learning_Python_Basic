```python
# __initial__ 있는 경우의 Class
class HamFamily():
    lastname = "Ham"
    def __init__(self, name):
        self.fullname = self.lastname + ' ' + name
    def travel(self, where):
        print('%s will trip to %s' %(self.fullname, where))
          
soyul = HamFamily("Soyul") 
soyul.travel("Busan")
```
