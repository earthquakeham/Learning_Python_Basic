```python
# 함수로 계산기 만들기
def num_sum(a,b):
    sum = a + b
    return sum
def num_mul(a,b):
    mul = a + b
    return mul

num_first = int(input("Input first number : "))
num_second = int(input("Input second number : "))
method = input("Input calculation method : ")

if method == "sum":
    print("Sum = ",(num_first + num_second))
elif method == "mul":
    print("Mul = ",(num_first * num_second))
else:
    print("Wrong input")
```

```python
# Class로 계산기 만들기
class Calculator:                            # Class 생성
    def data(self,first,second):             # method1 : Classd에 사용될 숫자 부여
        self.first = first
        self.second = second
    def sum(self):                           # method2 : 더하기                 
        result_sum = self.first + self.second
        return result_sum
    def mul(self):                           # method3 : 곱하기
        result_mul = self.first * self.second
        return result_mul
    
num_first = int(input("Input first number : "))
num_second = int(input("Input second number : "))
method = input("Input calculation method : ")

cal = Calculator()
cal.data(num_first,num_second)
if method == "sum":
    print("Sum = ",cal.sum())
elif method == "mul":
    print("Mul = ",cal.mul())
else:
    print("Wrong input")
```
