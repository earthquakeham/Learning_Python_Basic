```python
# Class HamFamily
class HamFamily():
    lastname = "Ham"
    def __init__(self, name):
        self.fullname = self.lastname + ' ' + name
    def travel(self, where):
        print('%s will trip to %s' %(self.fullname, where))

# Inheritance
class KimFamily(HamFamily):
    lastname = "Kim"
    
jihyun = KimFamily("Jihyun") 
jihyun.travel("Jeju")

# Inheritance and edit
class KimFamily(HamFamily):
    lastname = "Kim"
    def travel(self, where, day):   # day 추가 후 수정
        print('%s will trip to %s. @%s' %(self.fullname, where, day))  # day 추가 후 수정
    
jihyun = KimFamily("Jihyun") 
jihyun.travel("Jeju", "8/17") # day 추가 후 수정
```
