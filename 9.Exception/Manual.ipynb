```python
# try, except, else, finally
try:
    age = int(input("How old are you : "))
except:
    print("Wrong input")
else:
    if age <= 18:
        print("You can not use our website")
    else:
        print("Welcome")
finally:
    print("See you next time!")
```

```python
# 없는 file을 열려는 오류가 있으면 write mode로 file 생성
try:                               # 무언가를 수행
    f = open("exception.txt", 'r')
except FileNotFoundError:          # 어떤 error가 발생 시 대응방안
    print("No file, So open file as write mode.")
    f = open("exception.txt", 'w')
```
