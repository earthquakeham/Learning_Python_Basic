```python
# Python Introduction Step1

import os
import shutil

source_dir = r"C:\Users\lilin\Desktop\Test_Before"
dest_dir = r"C:\Users\lilin\Desktop\Test_After"

for file_name in os.listdir(source_dir):
    if file_name.endswith('.txt'):
        shutil.move(os.path.join(source_dir, file_name), os.path.join(dest_dir, file_name))

print(f"디렉토리 {source_dir}가 {dest_dir}로 복사되었습니다.")
```

```python
# Python Introduction Step3

import os
import shutil

source_dir = r"C:\Users\lilin\Desktop\Test_Before"
dest_dir = r"C:\Users\lilin\Desktop\Test_After"

for file_name in os.listdir(source_dir):
    if file_name.endswith('.txt'):
        shutil.move(os.path.join(source_dir, file_name), os.path.join(dest_dir+"\TXT", file_name))
    elif file_name.endswith('.bmp'):
        shutil.move(os.path.join(source_dir, file_name), os.path.join(dest_dir+"\BMP", file_name))
    elif file_name.endswith('.xlsx'):
        shutil.move(os.path.join(source_dir, file_name), os.path.join(dest_dir+"\Excel", file_name))
    elif file_name.endswith('.pptx'):
        shutil.move(os.path.join(source_dir, file_name), os.path.join(dest_dir+"\PPT", file_name))
    else:
        shutil.move(os.path.join(source_dir, file_name), os.path.join(dest_dir+"\Etc", file_name))

print(f"디렉토리 {source_dir}가 {dest_dir}로 복사되었습니다.")
```
